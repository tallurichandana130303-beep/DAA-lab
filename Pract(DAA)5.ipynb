```python
import time

# Taking user input
n = int(input("Enter the number of items: "))

weights = []
values = []

print("\nEnter the weights of the items:")
for i in range(n):
    weights.append(int(input(f"Weight of item {i + 1}: ")))

print("\nEnter the values of the items:")
for i in range(n):
    values.append(int(input(f"Value of item {i + 1}: ")))

capacity = int(input("\nEnter the maximum capacity of the knapsack: "))

# Start execution time
start_time = time.perf_counter()

# DP table
dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

# Dynamic Programming
for i in range(1, n + 1):
    for w in range(capacity + 1):

        if weights[i - 1] <= w:
            dp[i][w] = max(
                values[i - 1] + dp[i - 1][w - weights[i - 1]],
                dp[i - 1][w]
            )
        else:
            dp[i][w] = dp[i - 1][w]

# Maximum value
max_value = dp[n][capacity]

# Find selected items
w = capacity
selected_items = []

for i in range(n, 0, -1):
    if dp[i][w] != dp[i - 1][w]:
        selected_items.append(i)
        w -= weights[i - 1]

# End execution time
end_time = time.perf_counter()

execution_time = end_time - start_time

# Output
print("\n----- Knapsack Result -----")
print("Maximum value:", max_value)

print("Selected items:", end=" ")
for item in reversed(selected_items):
    print(item, end=" ")

print("\nExecution time: {:.8f} seconds".format(execution_time))
```
